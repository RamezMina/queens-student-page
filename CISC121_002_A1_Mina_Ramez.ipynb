```python
class Solution:
    def getSum(self, a, b):
        MASK = 0xFFFFFFFF
        MAX_INT = 0x7FFFFFFF
        while b != 0:
            carry = (a & b) & MASK
            a = (a ^ b) & MASK
            b = (carry << 1) & MASK # Keep looping till theres no carry.
        return a if a <= MAX_INT else ~(a ^ MASK) # Handle negative numbers

# Ai disclosure: I used ChatGPT (GPT-5) to help me fix the negative number issue, because my code only worked for positive numbers. And it helped me with the syntax, I was confused with some of the masking stuff.
```

```python
class Solution:
    def hammingWeight(self, n):
        count = 0
        while n > 0:
            if n % 2 == 1:
                count += 1
            n = n // 2
        return count

#https://leetcode.com/problems/number-of-1-bits/solutions/6942378/beats-100-00-2-approach-brute-force-optimal/
```

```python
class Solution:
    def reverse(self, x):
        # Determine the sign and work with the absolute value
        sign = -1 if x < 0 else 1
        x = abs(x)
        rev = 0
        # Reverse digits one by one
        while x != 0:
            digit = x % 10    # Get last digit
            rev = rev * 10 + digit    # Add digit to reversed number
            x //= 10    # Remove last digit from x
        rev *= sign  # Restore original sign
        # Check for 32-bit overflow
        if rev < -2**31 or rev > 2**31 - 1:
            return 0
        return rev

# AI Disclosure: I used ChatGPT (GPT-5) to help fix the sign logic and overflow handling.
```

```python
class Solution(object):
    def reverseString(self, s):
        left, right = 0, len(s) - 1
        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1

#https://leetcode.com/problems/reverse-string/solutions/7264290/reverse-string-100-beats-python/
```
